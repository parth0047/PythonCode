#Encoding (Not Recording)

```python
#ip/op Creation
ip = adv.drop('Sales',axis=1)
op = adv['Sales']
print(ip)
print(op)
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(ip,op,test_size=0.2,random_state=1)
```

```python
# shape all data set
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)


```

```python
#Scaling
#Standard Scaler Transform
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
#ML Algorithm
#Linear Regression
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)

```

```python
pred = lr.predict(x_test)
pred
```

```python
#Accuracy
from sklearn.metrics import r2_score, mean_squared_error
r2 = r2_score(y_test,pred)
mse = mean_squared_error(y_test,pred)
print(r2*100) #Accuracy
print(mse)    #Error in Model
```

```python
df = pd.DataFrame({'Actual':list(y_test),'Predicted':pred})
df
```

```python
#Implot - Linear Model Plot
sns.lmplot(x='Actual', y='Predicted', data=df,scatter_kws={'color':'green'},line_kws={'color':'red'})
plt.show()
#
```
